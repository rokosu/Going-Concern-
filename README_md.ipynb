# **README: Reproducibility Guide for**  
**"The Impact of Going Concern Uncertainties on Financial Statement Quality"**  

### **1. Dataset Description**  
**Source:**  
- **Archival Data:** 1,250 firm-year observations (2016–2020) from S&P 1500 (Compustat, Audit Analytics, ISS).  
- **Survey Data:** 500 professionals (250 auditors + 250 CFOs), stratified by firm size/industry.  

**Variables Included:**  
- `financial_quality.csv`: Dechow-Dichev accruals (dependent variable).  
- `disclosures.csv`: Management 8-K filings, auditor modifications.  
- `governance.csv`: Audit committee independence, CEO duality.  
- `distress.csv`: Altman Z-scores, materiality thresholds.  

---

### **2. Dataset Limitations**  
**a. Access Restrictions:**  
- **Compustat/ISS Data:** Requires subscription access. Provided here as aggregated/anonmyized due to licensing.  
- **Survey Data:** Raw responses excluded to protect confidentiality; only processed data included.  

**b. Missing Data Handling:**  
- Financial variables: Imputed via MICE (Multiple Imputation by Chained Equations).  
- Survey items: Used Full Information Maximum Likelihood (FIML).  

**c. Survivorship Bias:**  
- Sample limited to S&P 1500 firms (excludes private/small firms). Results may not generalize.  

**d. Temporal Scope:**  
- Data ends in 2020. Post-COVID trends (e.g., remote audits) are not captured.

### **4. Reproducing Results**  
**Step 1: Install Dependencies**  
```bash
pip install -r scripts/requirements.txt  # Includes linearmodels, statsmodels, pandas  
```
